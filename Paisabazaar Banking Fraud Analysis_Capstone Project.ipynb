#### Checking Outliers using IQR

Before drawing any charts I checked the numeric columns for outliers. The IQR rule is easy to apply: take the first quartile (Q1, the 25th percentile) and the third quartile (Q3, the 75th percentile). The difference between them is the IQR. Any value below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR` is counted as an outlier.

```python
# Outlier check using the IQR method
num_cols = ['Age', 'Annual_Income', 'Monthly_Inhand_Salary', 'Num_Bank_Accounts',
            'Num_Credit_Card', 'Interest_Rate', 'Num_of_Loan', 'Delay_from_due_date',
            'Num_of_Delayed_Payment', 'Num_Credit_Inquiries', 'Outstanding_Debt',
            'Credit_Utilization_Ratio', 'Total_EMI_per_month',
            'Amount_invested_monthly', 'Monthly_Balance']

rows = []
for col in num_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()

    rows.append([col, round(q1, 2), round(q3, 2), round(iqr, 2),
                 round(lower, 2), round(upper, 2), count,
                 round(count / len(df) * 100, 2)])

outliers = pd.DataFrame(rows, columns=['Column', 'Q1', 'Q3', 'IQR',
                                       'Lower bound', 'Upper bound',
                                       'Outliers', '% of rows'])
outliers.sort_values('Outliers', ascending=False)
```

```python
# How many rows are affected if we count every column together?
flag = pd.Series(False, index=df.index)
for i in range(len(outliers)):
    col = outliers.loc[i, 'Column']
    low = outliers.loc[i, 'Lower bound']
    high = outliers.loc[i, 'Upper bound']
    flag = flag | (df[col] < low) | (df[col] > high)

print('Rows with at least one outlier :', flag.sum())
print('As a share of the file         : {:.2f}%'.format(flag.mean() * 100))
print('')
print('Columns with no outliers at all:')
print(', '.join(outliers[outliers['Outliers'] == 0]['Column']))
print('')
print('Columns with the most outliers:')
top = outliers.sort_values('Outliers', ascending=False).head(3)
for i in range(3):
    print('  {:<24} {:>6} rows  ({:.1f}%)'.format(top.iloc[i]['Column'],
                                                 top.iloc[i]['Outliers'],
                                                 top.iloc[i]['% of rows']))
```

#### What I did about the outliers

I did not remove them, and I did not replace them with the quartile limits. There are two reasons.

The first is the size of the problem. 22,919 rows (22.9%) have at least one outlier, which is almost a quarter of the file. Deleting a quarter of the data to fix a few large values would cost more than it gains.

The second reason is what the outliers actually are. Every lower bound in the table is a negative number (for example -59,167 for `Annual_Income`), so nothing is being flagged as "too small". All the outliers are on the high side, and a customer with a high income or a large outstanding debt is not a data entry error. That is just a customer with a bigger loan. Removing those rows would delete the customers a bank cares about most.

The model in Chart 14 is a Random Forest, and it is not affected by large values in the same way as a mean is. A tree splits on order (is income above 50,000 or not), so one customer earning 180,000 does not drag the result the way it drags an average. In a notebook that used linear regression I would have capped the values first.

The boxplots in Chart 4 show where these outliers sit, so they are reported rather than hidden.

### Chart - 14 **Credit Score Prediction Model (Multivariate)**

### Building the model

```python
# Chart - 14 : predicting the credit score
from sklearn.model_selection import GroupShuffleSplit
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Columns used to predict the credit score
features = ['Age', 'Annual_Income', 'Num_Bank_Accounts', 'Num_Credit_Card',
            'Interest_Rate', 'Num_of_Loan', 'Delay_from_due_date',
            'Num_of_Delayed_Payment', 'Num_Credit_Inquiries', 'Outstanding_Debt',
            'Credit_Utilization_Ratio', 'Total_EMI_per_month',
            'Amount_invested_monthly', 'Monthly_Balance', 'Credit_Mix_Num']

X = df[features]
y = df['Credit_Score_Num']

print('Columns used for prediction :', len(features))
print('Rows                        :', len(X))
```

```python
# Split the data into a training set and a test set.
# I split by Customer_ID and not by row, because every customer appears 8 times in
# this file. If I split by row, the same customer would land in both the training and
# the test set, and the model would just remember that customer instead of learning
# the pattern. Splitting by customer keeps the test data honest.
splitter = GroupShuffleSplit(n_splits=1, test_size=0.25, random_state=42)
train_index, test_index = next(splitter.split(X, y, df['Customer_ID']))

X_train, X_test = X.iloc[train_index], X.iloc[test_index]
y_train, y_test = y.iloc[train_index], y.iloc[test_index]

print('Training rows :', len(X_train))
print('Testing rows  :', len(X_test))
print('Customers in both sets :',
      len(set(df['Customer_ID'].iloc[train_index]) & set(df['Customer_ID'].iloc[test_index])))
```

```python
# Train a Random Forest
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Predict on the test data
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
baseline = (y_test == 1).mean()

print('Model accuracy                        : {:.2f}%'.format(accuracy * 100))
print('Accuracy if we always said Standard   : {:.2f}%'.format(baseline * 100))
print('Improvement over that simple guess    : {:.2f} percentage points'.format(
      (accuracy - baseline) * 100))
```

```python
# Confusion matrix and feature importance
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

names = ['Poor', 'Standard', 'Good']
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=names, yticklabels=names, ax=axes[0])
axes[0].set_title('Confusion Matrix')
axes[0].set_xlabel('Predicted Credit Score')
axes[0].set_ylabel('Actual Credit Score')

importance = pd.Series(model.feature_importances_, index=features).sort_values()
importance.plot(kind='barh', color='steelblue', ax=axes[1])
axes[1].set_title('Which Columns the Model Used Most')
axes[1].set_xlabel('Importance')

plt.tight_layout()
plt.show()

print(importance.sort_values(ascending=False).round(3))
```

```python
# Result for each credit score group
print(classification_report(y_test, y_pred, target_names=['Poor', 'Standard', 'Good']))

print('How many each group got right:')
for i, name in enumerate(['Poor', 'Standard', 'Good']):
    right = cm[i][i]
    total = cm[i].sum()
    print('  {:<9} {:>5} of {:>6}  ({:.1f}%)'.format(name, right, total, right / total * 100))
```

#### **1. Why did you pick the specific chart?**

I used two charts together because they answer two different questions. The confusion matrix shows where the model is right and where it goes wrong, which is the part that matters if a person has to act on the result. The importance bar chart shows which columns the model actually used, and that is the check that the model is learning the right thing and not a coincidence.

I chose a Random Forest because it works with mixed data without scaling, it does not need the columns to be normally distributed, and it is easy to describe: it builds 100 decision trees and takes the majority vote. I could not use a simple linear model because the relationship between the columns and the score is not a straight line, which is what the curved pattern in the boxplots of Charts 7 and 8 shows.

#### **2. What is/are the insight(s) found from the chart?**

- The model gets 68.40% of the test rows right on customers it has never seen.
- If we always guessed Standard we would get 53.40%, so the model is 15 percentage points better than a guess that learns nothing.

| Group | Correct | Of total | Share |
|---|---|---|---|
| Poor | 4,790 | 7,246 | 66.1% |
| Standard | 9,545 | 13,349 | 71.5% |
| Good | 2,764 | 4,405 | 62.7% |

- The model is best at Standard, which is also the biggest group.
- Good is the hardest group. It finds 62.7% of them and misses about 4 in every 10.
- The most used columns are `Outstanding_Debt` (0.12), `Monthly_Balance` (0.09), `Credit_Utilization_Ratio` (0.09), `Interest_Rate` (0.09) and `Credit_Mix_Num` (0.09). Three of those five were also at the top of the heatmap in Chart 12, so two different methods picked the same columns.
- One thing surprised me. `Credit_Utilization_Ratio` comes third here even though its correlation with the score is only +0.046, which is almost zero. A Random Forest splits on many columns at once, so a column that carries very little on its own can still be used often inside the trees. That is why the correlation table and the importance chart should be read side by side and not one at a time.

#### **3. Will the gained insights help creating a positive business impact?**
**Are there any insights that lead to negative growth? Justify with specific reason.**

The business objective was a model that sorts customers into Good, Standard and Poor, and this does that at 68.40%, which is 15 points above the simple guess. The more useful result is that the model needs only 15 columns and no new data. The bank can run this on what it already collects.

There are three risks and all three are real.

The first is the Good group. The model finds 62.7% of the Good customers, so about 4 out of 10 are missed. If the bank used this model to approve loans automatically, it would turn away people who deserve a loan. Those are the customers a lender wants, so the loss is expensive as well as unfair.

The second is the accuracy itself. 68.40% means roughly 3 mistakes in every 10 rows. That is fine for sorting a queue, and it is not fine for a decision about money. A person should review the cases that sit near the border between two groups.

The third is about how the score is measured. Because every customer appears 8 times and most of the predictive columns stay the same in all 8 months, the model gives the same answer for all the months of one customer. So this model describes a customer and not a month. It cannot tell you that someone is getting worse, because nothing in the file changes when they do. A real early warning system needs data that moves, and that is a data collection problem and not a modelling problem.

# **Conclusion**

## Key Insights from Analysis

**Standard is the biggest group and Good is the smallest.** Standard has 53.2% of the records, Poor has 29.0% and Good only 17.8%. So the bank has a small number of low risk customers, and those are the ones worth protecting with better rates and pre approved offers.

**Credit Mix is the strongest single column.** Customers with a Bad mix are 60.1% Poor and only 1.5% Good, while customers with a Good mix are 48.9% Good and 16.0% Poor. That is a gap of more than 40 percentage points from one column that is already on the application form.

**Interest rate and payment delay separate the groups most cleanly.** The median interest rate is 21% for Poor and 7% for Good. The median delay is 27 days for Poor and 10 days for Good. Both columns are measured already, and the delay is the better one for early action because it is counted in days.

**Outstanding debt moves with the score as well.** A Poor customer owes about 1,945 and a Good customer about 737, so the median is 2.6 times higher in the Poor group.

**Income matters but much less than expected.** A Good customer earns 44,928 at the median and a Poor customer 32,065, which is 1.4 times more, but the boxes overlap heavily so income alone cannot decide a score. Age is even weaker at a correlation of +0.16.

**Occupation is not useful at all.** The highest and the lowest occupation differ by 3,445, which is 7.1%, while incomes inside a single occupation vary by around 38,289. The gap between groups is 11 times smaller than the spread inside them, so occupation should not be used to set loan amounts or to pick a target group.

**Annual Income and Monthly Inhand Salary are the same column twice.** Their correlation is 0.998, so one of them should be dropped before modelling.

**The IQR check found outliers but no errors.** 22,919 rows (22.9%) have at least one outlier, all of them on the high side, and six of the fifteen numeric columns have none. I kept them because a high income or a large loan is a real customer and not a mistake, and because a Random Forest is not disturbed by large values the way an average is.

**The data has two impossible values.** 5,868 rows have an age below 18 and 11,824 rows have a credit history longer than the person has been alive. Both were reported instead of deleted.

**The only fraud signal is in the two columns that were dropped.** 1,829 names, which is 18.1%, appear with between 2 and 6 SSNs, and those names sit behind 33.6% of the rows. A one line identity check catches this and needs no model.

**The model reaches 68.40%.** That is 15 points above the 53.40% baseline, and it uses the same columns the heatmap ranked highest. It is good enough to sort applications and not good enough to approve them alone, because it finds only 62.7% of the Good customers.

## Limitations

1. The credit score is not fixed for one customer. 58.3% of them change score during the 8 months, and most of the columns that predict the score do not change at all, so a model gives one answer for all 8 months of a person.
2. The data looks generated. `Credit_Utilization_Ratio` has 100,000 different values inside a narrow range of 20 to 50, and `Outstanding_Debt` never changes for a customer. Real bank data does not behave like this.
3. `Type_of_Loan` was not used. It holds 6,261 different combinations of loan types, and it would need to be split into separate Yes and No columns first. When I tested it separately it added almost nothing, because the number of loans already carries the same information.
4. The model is 68.40% accurate, which is not enough for automatic approval, and the Good group is the weakest part of it.
5. The fraud finding may not be real fraud. The repeated names are probably a side effect of how the file was built. The check is the useful part, not the specific names.
6. Outliers were kept and not capped, so any average in this notebook is pulled up by the high values. The medians are not affected.

## Next steps

1. Correct the two impossible values at the source, so that no customer is below 18 and no credit history is longer than the person's age.
2. Split `Type_of_Loan` into separate columns and test whether the loan types improve the model.
3. Use the most common score across three months for each customer instead of a single month, so the target is more stable.
4. Add the identity check as a fixed step in the loan process, before the score is calculated.
5. Try other models such as Logistic Regression and Gradient Boosting and compare them with the Random Forest. A Gradient Boosting model reached 68.91% on the same 15 columns when I tested it, so the difference is small and the limit is the data rather than the model.
